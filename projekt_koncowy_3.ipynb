```python
import pandas as pd

dane = pd.read_excel('Zrzut__z_sieci_2024_04_24.xlsx')

result = dane\
    .groupby(['description', 'chassis-module.name',
              'chassis-module.part-number',
              'chassis-module.model-number',
              'chassis-module.description'])\
    .size()\
    .reset_index(name='count')\
    .sort_values('count', ascending=False)\
    .drop_duplicates()\
    .reset_index(drop=True)

result.to_excel("card_types_in_specific_router.xlsx", index=False)
```
